<h1>Today's Main Project</h1>

```python
m_income = float(input("Enter your monthly income: "))

rent = int(input("Enter your monthly rent: "))
food_expenses = int(input("Enter your monthly food expenses: "))
transportation_expenses = int(input("Enter your monthly transportation expenses: "))

total_expenses = rent + food_expenses + transportation_expenses
savings = m_income - total_expenses
print("Your total monthly expenses are: $", total_expenses)
print("Your monthly savings are: $", savings)
```

<h1>
Create a Shopping Bill Calculator</h1>

```python
product_name = input("Enter the product name: ")
price_of_product = float(input("Enter the price of the product: "))
quantity_of_product = int(input("Enter the quantity of the product: "))
total_cost = price_of_product * quantity_of_product

print("Product Name:", product_name)
print("Price of Product: $", price_of_product)
print("Quantity of Product:", quantity_of_product)
print("The total cost of", quantity_of_product, product_name, "is: $", total_cost)
```
