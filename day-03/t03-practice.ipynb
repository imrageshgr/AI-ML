## Mini project

### Student Result Analyzer

```python
student_name = input("Enter student name here: ")
marks = int(input("Enter marks here (out of 100): "))

print("---### Final Display  ###---")
print("Student Name is:", student_name)
print("Marks:",marks)

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
    
```

### Movie Ticket Eligibility Checker

```python
name = input("Enter name here: ")
age = int(input("Enter your age: "))

print("--- ### Result ### ---")
print("Name:", name)
print("Age:", age)

if age < 5:
    print("Free Ticket")
elif age <= 17:
    print("Child Ticket")
elif age <= 59:
    print("Adult Ticket")
else:
    print("Senior Citizen Ticket")
```
