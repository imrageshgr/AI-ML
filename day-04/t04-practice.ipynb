## End-of-Day Challenge

### Electricity Bill Calculator

```python
units = float(input("Enter the number of units: "))

if units <= 100:
    bill = units * 2
elif units <= 200:
    bill = (100 * 2) + (units - 100) * 3
else:
    bill = (100 * 2) + (100 * 3) + (units - 200) * 5

print("The electricity bill is:", bill)
```
