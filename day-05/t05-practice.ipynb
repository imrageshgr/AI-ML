```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b != 0:
        return a / b
    else:
        return "Error: Division by zero is not allowed."

add_result = add(10, 5)
subtract_result = subtract(10, 5)
multiply_result = multiply(10, 5)
divide_result = divide(10, 5)

print("Addition Result:", add_result)
print("Subtraction Result:", subtract_result)
print("Multiplication Result:", multiply_result)
print("Division Result:", divide_result)
```

# Today's Main Project

Project: Smart Calculator

```python
first_number = int(input("Enter the first number: "))
second_number = int(input("Enter the second number: "))
operation = input("Enter the operation (+, -, *, /): ")

def perform_operation(num1, num2, op):
    if op == '+':
        return num1 + num2
    elif op == '-':
        return num1 - num2
    elif op == '*':
        return num1 * num2
    elif op == '/':
        if num2 != 0:
            return num1 / num2
        else:
            return "Error: Division by zero is not allowed."
    else:
        return "Error: Invalid operation."

perform_operation(first_number, second_number, operation)
```

# End-of-Day Challenge

Student Marks Analyzer

```python
student_name = input("Enter your name: ")
print("Tell three subjects marks")
sub1 = int(input("Enter marks for subject 1: "))
sub2 = int(input("Enter marks for subject 2: "))
sub3 = int(input("Enter marks for subject 3: "))

def total_marks(m1, m2, m3):
    return m1 + m2 + m3

def average_marks(m1, m2, m3):
    total = total_marks(m1, m2, m3)
    return total / 3

def grade(average):
    if average >= 90:
        return "A"
    elif average >= 75:
        return "B"
    elif average >= 50:
        return "C"
    else:
        return "Fail"

total = total_marks(sub1, sub2, sub3)
average = average_marks(sub1, sub2, sub3)
grade_result = grade(average)

print("Student Name:", student_name)
print("Total Marks:", total)
print("Average Marks:", average)
print("Grade:", grade_result)
```
