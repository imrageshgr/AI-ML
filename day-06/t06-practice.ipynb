## Main Project

#### Student Marks Analyzer

```python
numberOfStudents = int(input("Enter the number of students: "))

marks = []
for i in range(numberOfStudents):
    mark = int(input(f"Enter marks for student {i + 1}: "))
    marks.append(mark)

total = sum(marks)

def calculate_average(marks):
    total = sum(marks)
    average = total / len(marks)
    return average

highest_mark = max(marks)
lowest_mark = min(marks)

def count_passing_students(marks):
    passing_students = []
    failure_students = []
    for mark in marks:
        if mark >= 50:
            passing_students.append(mark)
        else:
            failure_students.append(mark)
    return len(passing_students), len(failure_students)

print("number of students:", numberOfStudents)
print("marks of students:", marks)
print("Total marks:", total)
average = calculate_average(marks)
print("Average marks:", average)
print("Highest mark:", highest_mark)
print("Lowest mark:", lowest_mark)
passing, failing = count_passing_students(marks)
print("Number of passing students:", passing)
print("Number of failing students:", failing)
```

## Today's independent challenge

Shopping Cart Analyzer

```python
number_of_products = int(input("Enter the number of products: "))
product_price = []

for i in range(number_of_products):
    price = float(input(f"Enter price for product {i + 1}: "))
    product_price.append(price)

total_price = sum(product_price)
highest_price = max(product_price)
lowest_price = min(product_price)
average_price = total_price / len(product_price)

above_500 = []
for price in product_price:
    if price > 500:
        above_500.append(price)

print("Number of products:", number_of_products)
print("Prices of products:", product_price)
print("Total price:", total_price)
print("Average price:", average_price)
print("Highest price:", highest_price)
print("Lowest price:", lowest_price)
print("Number of products above 500:", len(above_500))
```
