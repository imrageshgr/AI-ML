## Project

Student Performance Management System

```python
no_student = int(input("Enter the number of students: "))

student_names = []
all_marks = []

for i in range(no_student):
    name = input("Enter student name: ")
    student_names.append(name)

    marks = []

    for j in range(3):
        mark = int(input("Enter subject marks: "))
        marks.append(mark)

    all_marks.append(marks)


def total(marks):
    return sum(marks)


def average(marks):
    return sum(marks) / len(marks)


def grade(ave):
    if ave >= 90:
        return "A"
    elif ave >= 75:
        return "B"
    elif ave >= 50:
        return "C"
    else:
        return "F"


def result(ave):
    if ave >= 50:
        return "Pass"
    else:
        return "Fail"


for i in range(no_student):
    marks = all_marks[i]

    total_marks = total(marks)
    ave = average(marks)
    student_grade = grade(ave)
    student_result = result(ave)

    print("Student name:", student_names[i])
    print("Marks:", marks)
    print("Total marks:", total_marks)
    print("Average marks:", ave)
    print("Grade:", student_grade)
    print("Result:", student_result)
    print()
```

## Independent Challenge

Class Performance Analyzer

```python
number_of_students = int(input("Enter the number of students: "))

student_names = []
marks_list = []

for i in range(number_of_students):
    name = input("Enter student name: ")
    student_names.append(name)

    marks = []

    for j in range(3):
        mark = int(input("Enter subject marks: "))
        marks.append(mark)

    marks_list.append(marks)

def total(marks):
    return sum(marks)

def average(marks):
    return sum(marks) / len(marks)

def grade(ave):
    if ave >= 90:
        return "A"
    elif ave >= 75:
        return "B"
    elif ave >= 50:
        return "C"
    else:
        return "F"

def number_of_passed_students(marks_list):
    passed_count = 0
    for marks in marks_list:
        ave = average(marks)
        if ave >= 50:
            passed_count += 1
    return passed_count

def number_of_failed_students(marks_list):
    failed_count = 0
    for marks in marks_list:
        ave = average(marks)
        if ave < 50:
            failed_count += 1
    return failed_count

def highest_average(marks_list):
    highest_ave = 0
    for marks in marks_list:
        ave = average(marks)
        if ave > highest_ave:
            highest_ave = ave
    return highest_ave

def lowest_average(marks_list):
    lowest_ave = float('inf')
    for marks in marks_list:
        ave = average(marks)
        if ave < lowest_ave:
            lowest_ave = ave
    return lowest_ave

def class_average(marks_list):
    total_ave = 0
    for marks in marks_list:
        total_ave += average(marks)
    return total_ave / len(marks_list)

print("student results")
for i in range(number_of_students):
    marks = marks_list[i]

    total_marks = total(marks)
    ave = average(marks)
    student_grade = grade(ave)

    print("Student name:", student_names[i])
    print("Marks:", marks)
    print("Total marks:", total_marks)
    print("Average marks:", ave)
    print("Grade:", student_grade)
    print()

print("Number of passed students:", number_of_passed_students(marks_list))
print("Number of failed students:", number_of_failed_students(marks_list))
print("Highest average:", highest_average(marks_list))
print("Lowest average:", lowest_average(marks_list))
print("Class average:", class_average(marks_list))

```
